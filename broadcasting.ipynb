```python
#broadcasting
import numpy as np
```

```python
#senario 1
arr1=np.arange(8).reshape(2,4)
arr2=np.arange(8,16).reshape(2,4)
```

```python
arr1
```

```python
arr2
```

```python
arr1+arr2
```

```python
#senario 2
#if y=1 and x=n then also operation will take place
a=np.arange(9).reshape(3,3)
b=np.arange(3).reshape(1,3)
```

```python
a
```

```python
print(a,b)
```

```python
a+b
```

```python
#if y=1 and x!=n then also operation will not take place
a1=np.arange(9).reshape(3,3)
b1=np.arange(4).reshape(1,4)
```

```python
print(a1,b1)
```
